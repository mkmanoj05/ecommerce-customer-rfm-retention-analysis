```python
clean_df =df.copy()
# remove cancelled invoices
clean_df= clean_df[~clean_df["InvoiceNo"].astype(str).str.startswith("C")]
#remove customerId that is null
clean_df=clean_df[clean_df["CustomerID"].notna()]
# keep positive quantities
clean_df=clean_df[clean_df["Quantity"]>0]
# keep postive price 
clean_df=clean_df[clean_df["UnitPrice"]>0]
# make a new column revenue 
clean_df["Revenue"]=clean_df["Quantity"] * clean_df["UnitPrice"]
#reset index
clean_df=clean_df.reset_index(drop=True)

print("OriginalRows: ",len(df))
print("CleanRows: ",len(clean_df))
print("Rows Removed : ",len(df)-len(clean_df))
```

```python
#Check Cleaned Data 

print("Missing Customer Ids : ",clean_df["CustomerID"].isna().sum())
print("Negative Quantities :",(clean_df["Quantity"]<0).sum())
print("Negative or zero price :",(clean_df["UnitPrice"]<=0).sum())
print("Total Revenue : ",(clean_df["Revenue"]).sum())
print("Cancelled Invoices : ",(clean_df["InvoiceNo"].astype(str).str.startswith("C")).sum())
print("Unique Customers : ",clean_df["CustomerID"].nunique())
print("Unique Invoices : ",clean_df["InvoiceNo"].nunique())
```

```python
clean_df.to_csv("Online_retail_cleaned.csv",index=False)
```

```python
customer_summary = (
    clean_df.groupby("CustomerID")
                    .agg(
        total_revenue=("Revenue","sum"),
        total_quantity=("Quantity","sum"),
        total_transactions=("InvoiceNo","count"),
        unique_invoices=("InvoiceNo","nunique"),
        first_purchase=("InvoiceDate","min"),
        last_purchase=("InvoiceDate","max")
).reset_index() 
                   )

customer_summary.head()
```

```python
#Top 10 Customers
top_customers=(customer_summary.sort_values("total_revenue",ascending=False).head(10))
print(top_customers)
                                            
```

```python
customer_summary["customer_type"]=np.where(customer_summary["unique_invoices"]==1,"one_time_customer","repeat_customer")
print(customer_summary.head())
```

```python
print(customer_summary["customer_type"].value_counts())
```

```python
# What percentage of customers purchased only once?
print(customer_summary["customer_type"].value_counts(normalize=True)*100)
```

```python
repeat_customer_rate=((customer_summary["unique_invoices"]>1).sum() / len(customer_summary))*100

print(f"Repeat Customer Rate: {repeat_customer_rate :.2f}%")
```
